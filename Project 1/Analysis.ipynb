# PART A - Fundamentals


## 1. What is Data Analysis?

**Data Analysis** is the process of collecting, cleaning, organizing, and examining data to find useful information, patterns, and insights. It helps businesses and organizations make better decisions based on facts rather than assumptions.

### Steps of Data Analysis
- Data Collection
- Data Cleaning
- Data Transformation
- Data Visualization
- Interpretation of Results

### Conclusion
Data analysis helps convert raw data into meaningful information for decision-making.


## 2. Steps to Plan a Data Science Project

A Data Science project follows a structured approach:

### 1. Define Problem Statement
Understand the business problem clearly.

### 2. Collect Data
Gather data from databases, surveys, APIs, etc.

### 3. Data Cleaning
Remove missing values, duplicates, and errors.

### 4. Exploratory Data Analysis (EDA)
Analyze patterns and relationships using graphs and statistics.

### 5. Feature Engineering
Create useful input features for the model.

### 6. Model Selection
Choose suitable ML algorithms.

### 7. Model Training
Train the model using historical data.

### 8. Evaluation
Check performance using accuracy, precision, recall, etc.

### 9. Deployment
Deploy the model to real-world applications.

### 10. Monitoring
Continuously improve model performance.

## 4. What are Tensors?

A **tensor** is a mathematical structure used to store data in multiple dimensions. It is a generalization of scalars, vectors, and matrices.

### Types of Tensors

#### Scalar (0D Tensor)
Single number.

Example:
```python
5
```

#### Vector (1D Tensor)
List of numbers.

Example:
```python
[1, 2, 3]
```

#### Matrix (2D Tensor)
Rows and columns.

Example:
```python
[
 [1, 2],
 [3, 4]
]
```

#### Higher Dimensional Tensor (3D+)
Used for images, videos, and deep learning data.

---

## Tensor Explanation with NumPy

### Scalar Tensor
```python
import numpy as np

scalar = np.array(10)
print(scalar.ndim)
```

Output:
```python
0
```

### Vector Tensor
```python
vector = np.array([1, 2, 3])
print(vector.ndim)
```

Output:
```python
1
```

### Matrix Tensor
```python
matrix = np.array([[1, 2], [3, 4]])
print(matrix.ndim)
```

Output:
```python
2
```

### 3D Tensor
```python
tensor = np.array([
    [[1, 2], [3, 4]],
    [[5, 6], [7, 8]]
])

print(tensor.ndim)
```

Output:
```python
3
```

### Why Tensors are Important?
- Used in Machine Learning and Deep Learning
- Efficient for large data processing
- Essential in frameworks like TensorFlow and PyTorch

### Conclusion
Tensors are the backbone of modern Machine Learning and Deep Learning because they efficiently represent and process multi-dimensional data.

# Part D — Exploratory Data Analysis (EDA)

````python
# Visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns
````

## 8. Univariate Analysis

````python
# Histogram to see distribution of GenAI usage hours
plt.figure(figsize=(8,5))
plt.hist(df["Weekly_GenAI_Hours"], color="skyblue", edgecolor="black")
plt.title("Distribution of Weekly GenAI Hours", fontsize=14)
plt.xlabel("Hours")
plt.ylabel("Number of Students")
plt.grid(alpha=0.3)
plt.show()
````

## Interpretation
- This histogram shows how many hours students spend using GenAI weekly.
- Most students are concentrated in the medium usage range.
- Very low or very high usage students are fewer.

````python
# Count of burnout levels
sns.countplot(data=df, x="Burnout_Risk_Level", palette="pastel")
plt.title("Burnout Risk Level Distribution", fontsize=14)
plt.xticks(rotation=30)
plt.show()
````

## Interpretation
- This graph shows the number of students in each burnout risk category.
- It helps identify whether most students face low, medium, or high burnout.

## 9.Bivariate Analysis

````python
# Scatter plot to compare AI usage and GPA
plt.figure(figsize=(8,5))
sns.scatterplot(
    data=df,
    x="Weekly_GenAI_Hours",
    y="Post_Semester_GPA"
)
plt.title("AI Usage vs Post Semester GPA", fontsize=14)
plt.show()
````

## Interpretation
- This scatter plot shows the relationship between AI usage and academic performance.
- It helps determine whether higher AI usage improves GPA.

````python
# Relationship between dependency and burnout
sns.countplot(data=df, x="Perceived_AI_Dependency", hue="Burnout_Risk_Level", palette="Set2")
plt.title("AI Dependency vs Burnout", fontsize=14)
plt.xticks(rotation=45)
plt.show()
````

## Interpretation
- This graph compares AI dependency with burnout levels.
- Students with higher dependency may show increased burnout risk.

## 10.Multivairiate Analysis

````python
# Select numeric columns
num_cols = df.select_dtypes(include=["int64", "float64"])

# Correlation matrix
corr = num_cols.corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap="cool", linewidths=0.5)
plt.title("Correlation Heatmap", fontsize=14)
plt.show()
````

## Interpretation
- The heatmap displays correlation between numerical features.
- Positive values indicate direct relationship.
- Negative values indicate inverse relationship.

````python
import matplotlib.pyplot as plt

# Nice theme
sns.set_theme(style="darkgrid", palette="pastel")

# Custom pastel colors for 3 burnout levels
custom_palette = ["#89CFF0", "#6495ED", "#1E3A8A"]
# Pairplot
pair = sns.pairplot(
    df,
    vars=[
        "Pre_Semester_GPA",
        "Weekly_GenAI_Hours",
        "Post_Semester_GPA",
        "Skill_Retention_Score"
    ],
    hue="Burnout_Risk_Level",
    palette=custom_palette,
    diag_kind="hist",
    corner=True
)

pair.fig.suptitle("Pairplot of Student Performance and AI Usage", y=1.02)
plt.show()
````

## Interpretation
- Pair plot shows pairwise relationships among numeric columns.
- Useful for identifying trends, clusters, and outliers.
