```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv("training-master.csv")
test = pd.read_csv("testing-master.csv")
```

```python
# Step 1 

# First 5 Rows dekhna 
print("Training Data:")
print(train.head())

print("\nTesting Data:")
print(test.head())

# Rows aur Column count
print("\nTraining shape:" , train.shape)
print("Testing shape:" , test.shape)
```

#### Conclusion:
In Step 1, both training and testing datasets were successfully loaded. We checked the number of rows and columns and got a basic understanding of the dataset structure. This helps us start further EDA.

```python
# Step 2 - check Missing value + data types

# Training dataset info
print("Training Data Info:")
print(train.info())

# Testing dataset info
print("\nTesting Data Info:")
print(test.info())

# Missing value check
print("\nMissing Value In Training Data:")
print(train.isnull().sum())

print("\nMissing Value In Testing Data:")
print(test.isnull().sum())
```

#### Conclusion:

In Step 2, we checked the data types and missing values of both datasets. This helped us understand the structure of each column and identify whether any data cleaning is required

```python
# Step 3 - Statistical Summary

# Training Dataset Statistics
print("Trainig Data Summary:")
print(train.describe())

# Testing Dataset Statistics
print("\nTesting Data Summary:")
print(test.describe())
```

#### Conclusion:

In Step 3, we analyzed the statistical summary of numerical columns. This helped us understand the average, minimum, maximum, and spread of the data.

```python
# Step 4 - Distribution / Visulization of Numerical Columns

num_cols = ['Age', 'Tenure', 'Usage Frequency', 'Support Calls',
            'Payment Delay', 'Total Spend', 'Last Interaction']

for col in num_cols:
    train[col].hist(color="Lightblue")
    plt.title(col)
    plt.show()
```

Conclusion:

In Step 4, we visualized the numerical columns using histograms. This helped us understand the distribution of data and identify patterns or possible outliers in the dataset.

```python
# Step 5 - Categorical Columns Analysis

cat_cols = ['Gender', 'Subscription Type', 'Contract Length', 'Churn']

for col in cat_cols:
    print("\n", col)
    print(train[col].value_counts())

# For Graph
import matplotlib.pyplot as plt


# For Gender
counts = train['Gender'].value_counts()

plt.bar(counts.index, counts.values, color="Lightblue")
plt.title("Gender")
plt.show()


# for Subscription Type
counts = train['Subscription Type'].value_counts()

plt.bar(counts.index, counts.values, color="Lightpink")
plt.title("Subscription Type")
plt.show()


# For Contract Length
counts = train['Contract Length'].value_counts()

plt.bar(counts.index, counts.values, color="#bceabe")
plt.title("Contract Length")
plt.show()


# For Chuen
counts = train['Churn'].value_counts()

plt.bar(counts.index, counts.values, color="#c5a5b8")
plt.title("Churn")
plt.show()
```

#### Conclusion:

In Step 5, we analyzed all categorical columns using frequency counts and bar graphs. This helped us understand the distribution of different categories such as gender, subscription type, contract length, and churn status. It gave a clear view of the most common categories in the dataset.
