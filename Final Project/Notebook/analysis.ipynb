# 📘 Part D: Outlier Handling

---

## 7.1 Z-Score Method

````python
import numpy as np

# Select numerical columns
numeric_cols = ["age", "annual_income", "loan_amount", "credit_score"]

z_scores = np.abs(stats.zscore(df[numeric_cols]))
df_zscore = df[(z_scores < 3).all(axis=1)]

print("Original Shape:", df.shape)
print("After Z-Score:", df_zscore.shape)
````

### 💡 Interpretation

✅ The IQR method detects observations outside the range Q1 − 1.5×IQR and Q3 + 1.5×IQR. These extreme values were removed to improve data quality.

## 7.2 IQR Method

````python
df_iqr = df.copy()

for col in numeric_cols:

    Q1 = df_iqr[col].quantile(0.25)
    Q3 = df_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_iqr[(df_iqr[col] < lower) | (df_iqr[col] > upper)]

    print(f"\nColumn: {col}")
    print("Outliers:", outliers.shape[0])

    df_iqr = df_iqr[(df_iqr[col] >= lower) & (df_iqr[col] <= upper)]

print("\nOriginal Shape:", df.shape)
print("After IQR:", df_iqr.shape)
````

### 💡 Interpretation

✅ The Percentile method removes the lowest 1% and highest 1% of observations, reducing the influence of extreme values while preserving most of the data.

## 7.4 Winsorization Technique

````python
df_winsor = df.copy()

for col in numeric_cols:
    df_winsor[col] = winsorize(df_winsor[col], limits=[0.01, 0.01])

print(df_winsor.head())
````

### 💡Interpretation

✅ Winsorization replaces extreme values with the 1st percentile and 99th percentile values instead of deleting them. This reduces the effect of outliers while keeping all records in the dataset.

## 12. Apply Transformations

### 12.1 Function Transformer — Log Transform

````python
from sklearn.preprocessing import FunctionTransformer

log_transformer = FunctionTransformer(np.log1p)

df["annual_income_log"] = log_transformer.fit_transform(
    df[["annual_income"]]
)
````

### 12.2 Function Transformer — Reciprocal Transform

````python
reciprocal_transformer = FunctionTransformer(
    lambda x: 1 / (x + 1)
)

df["spending_ratio_reciprocal"] = reciprocal_transformer.fit_transform(
    df[["spending_ratio"]]
)
````

### 💡 Interpretation

 Reciprocal transformation helps reduce the effect of high values.
