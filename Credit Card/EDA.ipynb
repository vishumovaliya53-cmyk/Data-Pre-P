```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Step 1 - Load + head + count R/W

# Load
card = pd.read_csv("credit_card_data.csv")
print("File Loaded Successfully")

# Head
print("\nFirst 5 Rows:")
print(card.head())

# Count
print("\nCredit Card Shape:")
print(card.shape)
```

```python
# Step 2 - Check Missing value + data types

# Info
print("Data Info:")
print(card.info())

# Missing Value Check
print("\nMissing Value In Dataset:")
print(card.isnull().sum())
```

```python
# Step 3 - Statistical Summary

print("Data Summary:")
print(card.describe())
```

```python
# Step 4 - Numerical columns / visulization

num_cols = ['reports', 'age', 'income', 'share', 'expenditure', 'dependents',
       'months', 'majorcards']
    
for col in num_cols:
    card[col].hist(color="#13f635")
    plt.title(col)
    plt.show()

```

```python
# Step 5 - Categorical columns / visulization
cat_cols = card.select_dtypes(include=['object']).columns
print("\nCategorical Columns:")
print(cat_cols)

```
